## ③ Abstraire : une méthode de mise en équations

Dans les trois cas, on suit les mêmes étapes :

1. **Nommer les inconnues** et leurs unités.
2. **Traduire chaque contrainte par une équation.** Une loi, une conservation ou une donnée du texte devient une égalité.
3. **Compter** : autant d'équations indépendantes que d'inconnues ? Sinon, il manque une information, ou il y en a trop.
4. **Écrire le système sous forme matricielle** $A\mathbf{x} = \mathbf{b}$ : une ligne par équation, une colonne par inconnue.

### Le bijoutier

Inconnues : $x$ = masse d'or à 14 carats, $y$ = masse d'or à 22 carats, en grammes.

$$
\begin{cases}
x + y = 30 & \text{(masse totale)} \\
0{,}585\,x + 0{,}917\,y = 0{,}75 \times 30 = 22{,}5 & \text{(masse d'or pur)}
\end{cases}
$$

### La chimiste

Inconnues : $a$, $b$, $c$, $d$, les nombres de molécules dans

$$
a\,\mathrm{C_3H_8} + b\,\mathrm{O_2} \longrightarrow c\,\mathrm{CO_2} + d\,\mathrm{H_2O}
$$

Contrainte : rien ne se perd, rien ne se crée. Chaque **atome** est conservé.

$$
\begin{cases}
3a = c & \text{(carbone)} \\
8a = 2d & \text{(hydrogène)} \\
2b = 2c + d & \text{(oxygène)}
\end{cases}
$$

### L'électricien

Inconnues : $I_1$ (courant dans la résistance de $4\,\Omega$), $I_2$ (branche de $6\,\Omega$), $I_3$ (branche de $3\,\Omega$), en ampères.

Les lois de Kirchhoff fournissent les contraintes :

- Loi des nœuds : le courant qui arrive à l'embranchement repart dans les deux branches.
- Loi des mailles, premier tour : la pile, la résistance de $4\,\Omega$ et celle de $6\,\Omega$.
- Loi des mailles, deuxième tour : les deux branches parallèles ont la même tension.

$$
\begin{cases}
I_1 = I_2 + I_3 & \text{(nœud)} \\
12 = 4\,I_1 + 6\,I_2 & \text{(maille 1)} \\
6\,I_2 = 3\,I_3 & \text{(maille 2)}
\end{cases}
$$

### Une autre abstraction est-elle possible ?

Pour le bijoutier, on aurait pu n'utiliser qu'une inconnue : $y = 30 - x$, puis remplacer dans la seconde équation. Les deux abstractions donnent le même résultat. La forme matricielle a un avantage : elle se généralise sans effort à 3, 10 ou 1 000 inconnues.

## ④ Calculer

Pour un système carré dont la solution est unique, `np.linalg.solve(A, b)` renvoie le vecteur $\mathbf{x}$ tel que $A\mathbf{x} = \mathbf{b}$. Vous verrez au notebook 2.2 ce qu'il fait réellement.

### Le bijoutier

```python
import numpy as np

A_or = np.array([
    [1,     1    ],   # masse totale
    [0.585, 0.917],   # masse d'or pur
])
b_or = np.array([30, 0.75 * 30])

x_or = np.linalg.solve(A_or, b_or)
print(f"Or à 14 carats : {x_or[0]:.2f} g   Or à 22 carats : {x_or[1]:.2f} g")
```

### L'électricien

```python
# Une ligne par équation, colonnes : I1, I2, I3
A_circ = np.array([
    [1, -1, -1],   # nœud : I1 - I2 - I3 = 0
    [4,  6,  0],   # maille 1 : 4 I1 + 6 I2 = 12
    [0,  6, -3],   # maille 2 : 6 I2 - 3 I3 = 0
])
b_circ = np.array([0, 12, 0])

I = np.linalg.solve(A_circ, b_circ)
for nom, valeur in zip(["I1 (4 Ω)", "I2 (6 Ω)", "I3 (3 Ω)"], I):
    print(f"{nom} : {valeur:.3f} A")
```

### La chimiste

Trois équations, quatre inconnues : la matrice n'est pas carrée. Essayons tout de même.

```python
# Colonnes : a, b, c, d — chaque équation est écrite « ... = 0 »
A_chim = np.array([
    [3, 0, -1,  0],   # carbone   : 3a - c = 0
    [8, 0,  0, -2],   # hydrogène : 8a - 2d = 0
    [0, 2, -2, -1],   # oxygène   : 2b - 2c - d = 0
])
b_chim = np.zeros(3)

try:
    np.linalg.solve(A_chim, b_chim)
except np.linalg.LinAlgError as erreur:
    print("np.linalg.solve refuse :", erreur)
```

L'outil refuse : il n'accepte que les matrices carrées. Ce n'est pas un défaut du calcul, c'est une information sur le problème. Une équation de réaction peut toujours être multipliée par 2 ($2\,\mathrm{C_3H_8} + 10\,\mathrm{O_2} \to \dots$) : il y a donc une **infinité** de solutions. Pour en choisir une, les chimistes ajoutent une convention : le plus petit nombre entier possible devant le propane, soit $a = 1$.

Cette convention est une quatrième équation, et le système devient carré.

```python
A_chim4 = np.vstack([A_chim, [1, 0, 0, 0]])   # convention : a = 1
b_chim4 = np.array([0, 0, 0, 1])

a_, b_, c_, d_ = np.linalg.solve(A_chim4, b_chim4)
print(f"{a_:.0f} C3H8 + {b_:.0f} O2  ->  {c_:.0f} CO2 + {d_:.0f} H2O")
```

## Corrigé

*À n'ouvrir qu'après avoir cherché.*

```python
def melange(titre_1, titre_2, titre_cible, masse):
    if titre_1 == titre_2:
        return None                                   # matrice singulière : les deux lignes sont proportionnelles
    m1, m2 = np.linalg.solve([[1, 1], [titre_1, titre_2]], [masse, titre_cible * masse])
    if m1 < 0 or m2 < 0:
        return None                                   # titre visé hors de l'intervalle : il faudrait une masse négative
    return round(float(m1), 2), round(float(m2), 2)

print(melange(0.585, 0.917, 0.75, 30))
print(melange(0.59, 0.92, 0.75, 30))                  # or un peu plus pur
print(melange(0.585, 0.917, 0.95, 30))                # 23 carats : impossible avec ces chutes
print(melange(0.75, 0.75, 0.75, 30))                  # deux chutes identiques
```
