## Corrigé

*À n'ouvrir qu'après avoir cherché.*

```python
print("Test :", np.cumsum([1, 2, 3, 4]))          # [ 1  3  6 10]

cumul = np.cumsum(epargne)
verse = np.where(e >= 200, e + 50, e)
cumul_bonus = np.cumsum(verse)
dates = np.linspace(0, 12, 7)

print("Cumul sans bonus :", cumul)
print("Montants versés  :", verse)
print("Cumul avec bonus :", cumul_bonus)
print("Dates            :", dates)

for nom, c in [("sans bonus", cumul), ("avec bonus", cumul_bonus)]:
    atteint = np.nonzero(c >= 1800)[0]
    print(f"{nom} : objectif atteint au mois {atteint[0] + 1}" if atteint.size else f"{nom} : objectif non atteint en {len(c)} mois")
print("Votre pari :", pari_mois_objectif)
print("Vérification :", cumul[-1] == sum(epargne))
```
