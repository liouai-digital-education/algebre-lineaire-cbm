## ③ Abstraire : composer, c'est multiplier

Appliquer $B$, puis $A$ au résultat, envoie $\mathbf{x}$ sur $A(B\mathbf{x})$. Le produit matriciel est **défini** pour que cet enchaînement soit une seule matrice :

$$
A(B\mathbf{x}) = (AB)\,\mathbf{x}
$$

![Schéma : x est envoyé par B sur Bx, puis par A sur A(Bx) ; en une seule fois par le produit AB](../../assets/schemas/schema-composition.svg)

Pour des matrices $2 \times 2$, le coefficient en ligne $i$ et colonne $j$ du produit est le produit scalaire de la ligne $i$ de $A$ par la colonne $j$ de $B$ :

$$
\begin{pmatrix} a & b \\ c & d \end{pmatrix}
\begin{pmatrix} e & f \\ g & h \end{pmatrix}
=
\begin{pmatrix} ae + bg & af + bh \\ ce + dg & cf + dh \end{pmatrix}
$$

## ④ Calculer

```python
import numpy as np
import matplotlib.pyplot as plt

# Un point est une COLONNE : un ensemble de n points est un tableau de forme (2, n).
# Ainsi, appliquer la matrice A à tous les points s'écrit simplement A @ points.

# La lettre F : asymétrique, elle montre les rotations ET les retournements
F = np.array([[0, 0, 2, 2, 0.5, 0.5, 1.5, 1.5, 0.5, 0.5, 0],
              [0, 3, 3, 2.5, 2.5, 1.7, 1.7, 1.2, 1.2, 0, 0]])

def grille(ax, A=np.eye(2), n=4, **style):
    # Trace l'image par A des droites x = k et y = k, pour k entre -n et n
    t = np.linspace(-n, n, 2)
    for k in range(-n, n + 1):
        for ligne in (np.array([t, np.full(2, k)]), np.array([np.full(2, k), t])):
            image = A @ ligne
            ax.plot(*image, lw=0.6, color=style.get("couleur", "#D5DBDE"), zorder=0)

def dessiner(ax, A, titre="", limite=4.5):
    # Grille transformée, lettre F avant (gris) et après (vert), images des vecteurs de base
    grille(ax, A)
    ax.fill(*F, color=LIOUAI["gris"], alpha=0.35, lw=0)
    image = A @ F
    ax.fill(*image, color=LIOUAI["vert"], alpha=0.55)
    ax.plot(*image, color=LIOUAI["vert_texte"], lw=1.5)
    for colonne, couleur in [(A[:, 0], LIOUAI["vert_texte"]), (A[:, 1], LIOUAI["navy"])]:
        ax.annotate("", xy=colonne, xytext=(0, 0),
                    arrowprops=dict(arrowstyle="-|>", lw=2, color=couleur, mutation_scale=14))
    ax.set_xlim(-limite, limite); ax.set_ylim(-limite, limite); ax.set_aspect("equal")
    ax.axhline(0, color=LIOUAI["gris"], lw=0.8); ax.axvline(0, color=LIOUAI["gris"], lw=0.8)
    ax.set_xticks([]); ax.set_yticks([])
    ax.set_title(titre, fontsize=11)

# Le catalogue des transformations
def rotation(theta_deg):
    t = np.radians(theta_deg)
    return np.array([[np.cos(t), -np.sin(t)], [np.sin(t), np.cos(t)]])

def homothetie(k):
    return k * np.eye(2)

def dilatation(kx, ky):
    return np.array([[kx, 0], [0, ky]])

def cisaillement(k):
    return np.array([[1, k], [0, 1]])

def projection_sur_x():
    return np.array([[1, 0], [0, 0]])

def symetrie_axe_y():
    return np.array([[-1, 0], [0, 1]])
```

```python
R = rotation(90)
D = dilatation(2, 1)

noe = D @ R      # d'abord R, ensuite D
sarah = R @ D    # d'abord D, ensuite R
print("Noé   (D R) :", np.round(noe).astype(int).tolist())
print("Sarah (R D) :", np.round(sarah).astype(int).tolist())
print("Identiques ?", np.allclose(noe, sarah))

fig, axes = plt.subplots(1, 2, figsize=(9, 4.5))
dessiner(axes[0], noe, "Noé : pivoter puis étirer")
dessiner(axes[1], sarah, "Sarah : étirer puis pivoter")
plt.tight_layout(); plt.show()
```

### Quelles paires commutent ?

On dit que $A$ et $B$ **commutent** si $AB = BA$. Testons toutes les paires d'un petit catalogue.

```python
catalogue = {
    "rotation 30°": rotation(30),
    "rotation 75°": rotation(75),
    "homothétie × 2": homothetie(2),
    "étirement x × 2": dilatation(2, 1),
    "étirement y × 3": dilatation(1, 3),
    "inclinaison": cisaillement(0.5),
    "miroir": symetrie_axe_y(),
}
noms = list(catalogue)
commute = np.array([[np.allclose(catalogue[a] @ catalogue[b], catalogue[b] @ catalogue[a]) for b in noms] for a in noms])

fig, ax = plt.subplots(figsize=(7, 6))
ax.imshow(commute, cmap="liouai", vmin=0, vmax=1.6)
ax.set_xticks(range(len(noms)), noms, rotation=45, ha="right"); ax.set_yticks(range(len(noms)), noms)
for i in range(len(noms)):
    for j in range(len(noms)):
        ax.text(j, i, "oui" if commute[i, j] else "non", ha="center", va="center", fontsize=9,
                color="white" if commute[i, j] else LIOUAI["navy"])
ax.set_title("AB = BA ?")
plt.tight_layout(); plt.show()
```

### Deux miroirs font une rotation

```python
def miroir(angle_deg):
    # Symétrie par rapport à la droite passant par l'origine et faisant l'angle donné avec l'axe x
    t = np.radians(2 * angle_deg)
    return np.array([[np.cos(t), np.sin(t)], [np.sin(t), -np.cos(t)]])

S1, S2 = miroir(0), miroir(30)
produit = S2 @ S1
angle = np.degrees(np.arctan2(produit[1, 0], produit[0, 0]))
print("S2 S1 =", np.round(produit, 3).tolist())
print(f"C'est la rotation d'angle {angle:.0f}°")

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
dessiner(axes[0], S1, "Miroir d'axe 0°")
dessiner(axes[1], S2, "Miroir d'axe 30°")
dessiner(axes[2], produit, "Le premier, puis le second")
plt.tight_layout(); plt.show()
```

## Corrigé

*À n'ouvrir qu'après avoir cherché.*

Deux matrices diagonales commutent :

$$
\begin{pmatrix} a & 0 \\ 0 & d \end{pmatrix} \begin{pmatrix} e & 0 \\ 0 & h \end{pmatrix} = \begin{pmatrix} ae & 0 \\ 0 & dh \end{pmatrix} = \begin{pmatrix} e & 0 \\ 0 & h \end{pmatrix} \begin{pmatrix} a & 0 \\ 0 & d \end{pmatrix}
$$

Formules d'addition, lues dans la première colonne de $R_\beta R_\alpha = R_{\alpha+\beta}$ :

$$
\cos(\alpha+\beta) = \cos\alpha\cos\beta - \sin\alpha\sin\beta, \qquad \sin(\alpha+\beta) = \sin\alpha\cos\beta + \cos\alpha\sin\beta
$$

```python
def commutent(A, B, tol=1e-12):
    # Comparer les matrices revient à comparer les images de e1 ET de e2 (les colonnes) : deux points suffisent
    return np.allclose(A @ B, B @ A, atol=tol)

print(commutent(D2, C), commutent(rotation(20), rotation(50)), commutent(homothetie(3), cisaillement(2)))
for a1, a2 in [(0, 30), (10, 55), (20, 110)]:
    P = miroir(a2) @ miroir(a1)
    print(f"miroirs {a1}° puis {a2}° -> rotation de {np.degrees(np.arctan2(P[1, 0], P[0, 0])):.0f}° = 2 × ({a2} - {a1})")
```
