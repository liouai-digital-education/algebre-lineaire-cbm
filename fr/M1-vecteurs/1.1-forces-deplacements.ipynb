## ③ Abstraire : le concept de vecteur

Un **vecteur** décrit un déplacement : une longueur et une direction. Dans un repère (axe $x$ vers l'**est**, axe $y$ vers le **nord**), il s'écrit avec deux **coordonnées**. On note les vecteurs en gras, $\mathbf{u}$ ; au lycée, on les note souvent $\vec{u}$.

- 2 km vers l'ouest : $(-2\,;\,0)$.
- 1,5 km vers le sud : $(0\,;\,-1{,}5)$.
- 3 km vers le nord-est (direction à 45°) : $(3\cos 45^\circ\,;\,3\sin 45^\circ) \approx (2{,}12\,;\,2{,}12)$.

Trois opérations suffisent pour la plupart des problèmes :

| Opération | Sens concret | En coordonnées |
|---|---|---|
| **Somme** $\mathbf{u} + \mathbf{v}$ | enchaîner deux déplacements | on additionne les coordonnées |
| **Multiplication par un nombre** $k\mathbf{u}$ | aller $k$ fois plus loin ; $k$ négatif : faire demi-tour | on multiplie chaque coordonnée par $k$ |
| **Norme** $\lVert \mathbf{u} \rVert$ | la longueur du déplacement | $\sqrt{x^2 + y^2}$, par le théorème de Pythagore |

## ④ Calculer

### La course d'orientation

```python
import numpy as np
import matplotlib.pyplot as plt

def vecteur(longueur, angle_degres):
    # Déplacement de longueur donnée, dans la direction angle_degres (0° = est, 90° = nord)
    t = np.radians(angle_degres)
    return longueur * np.array([np.cos(t), np.sin(t)])

etapes = [vecteur(3, 45), vecteur(2, 180), vecteur(1.5, 270)]   # nord-est, ouest, sud
for e in etapes:
    print(np.round(e, 3))
```

```python
position = sum(etapes)                     # somme des trois vecteurs
distance = np.linalg.norm(position)        # sa norme
retour = -position                         # le chemin du retour : le vecteur opposé
angle_retour = np.degrees(np.arctan2(retour[1], retour[0]))

print("Position finale (est, nord) en km :", position.round(3))
print(f"Distance au départ : {distance:.2f} km")
print(f"Direction du retour : {angle_retour:.0f}° (0° = est, 90° = nord, -90° = sud)")
```

**Expliquer visuellement.** Le dessin des vecteurs « bout à bout » montre pourquoi la somme est le déplacement total.

```python
fig, ax = plt.subplots(figsize=(5, 5))
depart = np.zeros(2)
for e, nom in zip(etapes, ["3 km NE", "2 km O", "1,5 km S"]):
    ax.annotate("", xy=depart + e, xytext=depart, arrowprops=dict(arrowstyle="->", lw=2, color=LIOUAI["vert_texte"]))
    ax.text(*(depart + e / 2 + np.array([0.08, 0.08])), nom, color=LIOUAI["vert_texte"])
    depart = depart + e
ax.annotate("", xy=position, xytext=(0, 0), arrowprops=dict(arrowstyle="->", lw=2.5, color=LIOUAI["navy"]))
ax.text(*(position / 2 + np.array([0.1, -0.15])), "somme", color=LIOUAI["navy"], weight="bold")
ax.plot(0, 0, "ko"); ax.text(0.05, -0.2, "départ")
ax.set_xlabel("est (km)"); ax.set_ylabel("nord (km)")
ax.set_aspect("equal"); ax.grid(alpha=0.3)
ax.set_xlim(-0.5, 2.6); ax.set_ylim(-0.5, 2.6)
plt.show()
```

### La rivière

Les vitesses aussi sont des vecteurs. La vitesse réelle du kayak est la **somme** de sa vitesse propre (celle qu'il aurait en eau calme) et de la vitesse du courant :

$$
\mathbf{v}_{\text{réelle}} = \mathbf{v}_{\text{propre}} + \mathbf{c}
$$

![Schéma : le kayak qui pointe vers le nord dérive vers l'est ; s'il vise l'amont avec un angle thêta, il traverse tout droit](../../assets/schemas/schema-kayak.svg)

```python
largeur = 300                    # m
courant = np.array([1.0, 0.0])   # m/s, vers l'est
vitesse_propre = 2.0             # m/s

# Cas 1 : il pointe droit vers le nord
v_reelle = vecteur(vitesse_propre, 90) + courant
duree = largeur / v_reelle[1]                 # temps pour franchir 300 m vers le nord
derive = v_reelle[0] * duree                  # distance parcourue vers l'est pendant ce temps
print(f"Vitesse réelle : {v_reelle.round(2)} m/s ; traversée en {duree:.0f} s ; dérive de {derive:.0f} m vers l'est")
```

**Arriver juste en face.** Il faut que la composante est de la vitesse réelle soit nulle : le kayak doit pointer légèrement vers l'amont (vers l'ouest) pour compenser le courant. Si $\theta$ est l'angle entre sa direction et le nord, la composante ouest de sa vitesse propre vaut $2\sin\theta$ (en m/s). Elle doit compenser exactement le courant :

$$
2\sin\theta = 1 \quad\Longleftrightarrow\quad \theta = \arcsin\frac{1}{2} = 30^\circ
$$

```python
def cap_pour_traverser(vitesse_propre, vitesse_courant):
    # Angle (en degrés, vers l'amont) à prendre par rapport à la perpendiculaire aux rives
    rapport = vitesse_courant / vitesse_propre
    if rapport >= 1:
        return None                          # impossible : le courant est trop fort
    return np.degrees(np.arcsin(rapport))

theta = cap_pour_traverser(2.0, 1.0)
v = vecteur(2.0, 90 + theta) + courant       # pointer à theta degrés à l'ouest du nord
print(f"Cap : {theta:.0f}° vers l'amont ; vitesse réelle {v.round(3)} m/s ; traversée en {largeur / v[1]:.0f} s")
```

```python
def bilan_parcours(etapes):
    pos = sum(vecteur(l, ang) for l, ang in etapes)
    d = float(np.linalg.norm(pos))
    if d < 1e-9:
        return 0.0, None                     # déjà au départ : pas de direction de retour
    return round(d, 3), round(float(np.degrees(np.arctan2(-pos[1], -pos[0]))), 1)

print(bilan_parcours([(3, 45), (2, 180), (1.5, 270)]))
print(bilan_parcours([(1, 0), (1, 90), (1, 180), (1, 270)]))   # un carré : retour au départ, distance 0
```
